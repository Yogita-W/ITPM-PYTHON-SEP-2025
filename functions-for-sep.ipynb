```python
#WAp to check given number is even or odd using function.
def Even_number(num):
    if num%2==0:
        return f"{num} is Even number"

    else : 
        return f"{num} is Odd"

Even_number(20)
```

Different Ways to Assign Agruments to function.

```python
#Assigning single parameter or Arg.
def Number(num):
    return num

Number(20)
```

```python
#Multiple parameter
def Number(num1,num2):
    return f"{num1} and {num2}"

Number(20,30)
```

```python
def Marks(*num):
    return num

print(Marks(10))
print(Marks(10,20))
print(Marks(10,20,30))
print(Marks(10,20,30,40))
print(Marks([1,2,3,4],[6,7,8]))
```

```python
def Numbers(a,b,c):
    return b 
print(Numbers(10,20,30))
print(Numbers(30,10,40))
```

```python
#key-values 
def Numbers(a,b,c):
    return b 
print(Numbers(b=10,a=20,c=30))
```

```python
def Std_info(Name,Address,Marks="Absent"):
    return Name,Address,Marks

print(Std_info("Pari","Pune",90))
print(Std_info("Jay","Mumbai"))
```
